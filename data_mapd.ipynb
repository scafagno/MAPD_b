```python
import numpy as np
import pandas as pd
from pathlib import Path
from tqdm.notebook import tqdm

def decode_file(file_path):
    data = np.fromfile(file_path, dtype=np.uint64)
    df = pd.DataFrame({
        "TDC":   (data >> 0)  & 0x1F,
        "BX":    (data >> 5)  & 0xFFF,
        "ORBIT": (data >> 17) & 0xFFFFFFFF,
        "CHAN":  (data >> 49) & 0x1FF,
        "FPGA":  (data >> 58) & 0x7,
        "HEAD":  (data >> 61) & 0x7
    })
    
    return df
dfs = {} 

for i in tqdm(range(80)):
    file_path = Path(f"/home/samu/data_mapd/data_{i:06d}.dat")
    dfs[f"dtqdm(f_{i}"] = decode_file(file_path)
    #print(len(dfs)) 
    
#print(len(dfs))

```

```python
import numpy as np
import pandas as pd
from pathlib import Path

def decode_file(file_path):
    data = np.fromfile(file_path, dtype=np.uint64)
    df = pd.DataFrame({
        "TDC":   (data >> 0)  & 0x1F,
        "BX":    (data >> 5)  & 0xFFF,
        "ORBIT": (data >> 17) & 0xFFFFFFFF,
        "CHAN":  (data >> 49) & 0x1FF,
        "FPGA":  (data >> 58) & 0x7,
        "HEAD":  (data >> 61) & 0x7
    })
    
    return df
dfs = [] 

for i in tqdm(range(80)):
    file_path = Path(f"/home/samu/data_mapd/data_{i:06d}.dat")
    df = decode_file(file_path)
    dfs.append(df)
 #   print(len(dfs)) 
    
#print(len(dfs))

```
